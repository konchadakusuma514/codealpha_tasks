## Business Question 1

### Which country experienced the highest number of cyber attacks?

This visualization analyzes the distribution of cyber attacks across different countries. It helps identify which countries reported the highest number of cyber attack incidents during the period 2015–2024.

```python
plt.figure(figsize=(10,5))

sns.countplot(data=df, x="Country", order=df["Country"].value_counts().index)

plt.title("Number of Cyber Attacks by Country")
plt.xticks(rotation=45)
plt.show()
```

Countries with taller bars have experienced a greater number of cyber attacks compared to others. This indicates that cyber threats are not equally distributed across all countries.

### Business Insight

Countries reporting a higher number of cyber attacks should strengthen their cybersecurity infrastructure, improve employee awareness programs, and invest in advanced threat detection systems. Governments and organizations in these countries should prioritize cybersecurity policies to reduce future incidents.

# Business Question 10

## Which cyber attack type results in the highest average financial loss?

### Objective

This analysis aims to compare the average financial loss caused by different cyber attack types. Understanding the financial impact of each attack type helps organizations identify the most costly cyber threats and prioritize investments in cybersecurity measures.

```python
plt.figure(figsize=(10,5))

avg_loss = df.groupby("Attack Type")["Financial Loss (in Million $)"].mean().sort_values(ascending=False)

sns.barplot(
    x=avg_loss.index,
    y=avg_loss.values
)

plt.title("Average Financial Loss by Attack Type", fontsize=15, fontweight='bold')
plt.xlabel("Attack Type")
plt.ylabel("Average Financial Loss (Million $)")
plt.xticks(rotation=45)

plt.show()
```

### Observation

The bar chart compares the average financial loss associated with different cyber attack types. It highlights which attack types have a greater financial impact on organizations, allowing for easy comparison of the economic consequences of each threat.

### Business Insight

Not all cyber attacks have the same financial impact. Some attack types may occur less frequently but result in significantly higher financial losses. Organizations should prioritize preventive measures against these high-impact attacks to reduce financial risks and protect critical business assets.

### Recommendation

- Prioritize security investments for attack types that result in the highest financial losses.
- Implement advanced threat detection and continuous monitoring systems.
- Develop comprehensive incident response and disaster recovery plans to minimize financial damage.
- Regularly back up critical business data to ensure quick recovery after an attack.
- Conduct periodic cyber risk assessments to identify high-impact threats and strengthen security controls accordingly.

# Business Question 11

## Which cyber attack type requires the longest average incident resolution time?

### Objective

This analysis aims to compare the average incident resolution time for different cyber attack types. Understanding which attacks require more time to resolve helps organizations identify complex threats, improve incident response planning, and optimize cybersecurity operations.

```python
plt.figure(figsize=(10,5))

avg_time = (
    df.groupby("Attack Type")["Incident Resolution Time (in Hours)"]
    .mean()
    .sort_values(ascending=False)
)

sns.barplot(
    x=avg_time.index,
    y=avg_time.values
)

plt.title("Average Resolution Time by Attack Type", fontsize=15, fontweight="bold")
plt.xlabel("Attack Type")
plt.ylabel("Average Resolution Time (Hours)")
plt.xticks(rotation=45)

plt.show()
```

### Observation

The bar chart compares the average incident resolution time across different cyber attack types. Attack types with taller bars require more time to investigate, contain, and recover from, indicating greater complexity or operational impact.

### Business Insight

Cyber attacks that require longer resolution times can lead to increased operational downtime, higher recovery costs, and greater business disruption. Understanding these attack types enables organizations to strengthen their incident response capabilities and allocate resources more effectively.

### Recommendation

- Develop a well-defined incident response plan for high-impact cyber attacks.
- Use Security Information and Event Management (SIEM) tools for faster threat detection.
- Automate repetitive incident response tasks where possible.
- Conduct regular incident response drills to improve team readiness.
- Continuously monitor and analyze incident response metrics to reduce average resolution time.

# Business Question 13

## Is there a relationship between financial loss and incident resolution time?

### Objective

This analysis examines the relationship between the financial loss caused by cyber attacks and the time required to resolve those incidents. Understanding this relationship helps determine whether attacks with greater financial impact generally require longer recovery and response times.

```python
plt.figure(figsize=(8,5))

sns.regplot(
    data=df,
    x="Financial Loss (in Million $)",
    y="Incident Resolution Time (in Hours)",
    scatter_kws={"alpha":0.6},
    line_kws={"color":"red"}
)

plt.title("Financial Loss vs Incident Resolution Time", fontsize=15, fontweight="bold")
plt.xlabel("Financial Loss (Million $)")
plt.ylabel("Incident Resolution Time (Hours)")

plt.show()
```

### Observation

The scatter plot visualizes the relationship between financial loss and incident resolution time for individual cyber attack incidents.

Each point represents one cyber attack. If the points appear to follow an upward trend, it suggests that incidents causing higher financial losses may also require longer resolution times. If the points are widely scattered without a clear pattern, the relationship between these variables is weak or minimal.

### Business Insight

Understanding the relationship between financial loss and incident resolution time helps organizations assess the operational impact of cyber attacks. If more costly incidents consistently require longer recovery periods, organizations should focus on improving incident response capabilities to minimize downtime, financial losses, and business disruption.

### Recommendation

- Strengthen incident response teams to reduce recovery time for critical cyber attacks.
- Invest in automated detection and response solutions to accelerate incident handling.
- Prioritize rapid containment of attacks with the potential to cause significant financial damage.
- Continuously monitor cybersecurity performance indicators such as Mean Time to Detect (MTTD) and Mean Time to Respond (MTTR).
- Use predictive analytics to identify incidents that may require extended recovery efforts and allocate resources proactively.
